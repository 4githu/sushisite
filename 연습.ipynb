```python
import sqlite3

DB_PATH = "calendar.db"


def get_connection():
    conn = sqlite3.connect(DB_PATH)
    conn.row_factory = sqlite3.Row
    return conn
```

```python
import json

class Event:
    def __init__(self, id, title, start_time, end_time, group_id=None, is_all_day=False, status=0, type=0, repeat_group_id=None, color="#FFFFFF", memo="", metadata=None):
        self.id = id
        self.title = title
        self.start_time = start_time
        self.end_time = end_time
        self.group_id = group_id
        self.is_all_day = is_all_day
        self.status = status
        self.type = type
        self.repeat_group_id = repeat_group_id
        self.color = color
        self.memo = memo
        self.metadata = metadata

    def make(self, row):
        return self(
            id=row["id"],
            title=row["title"],
            start_time=row["start_time"],
            end_time=row["end_time"],
            group_id=row["group_id"],
            is_all_day=bool(row["is_all_day"]),
            status=row["status"],
            type=row["type"],
            repeat_group_id=row["repeat_group_id"],
            color=row["color"],
            memo=row["memo"],
            metadata=json.loads(row["metadata"])
        )

class Group:
    def __init__(self, id, name, parent_group_id=None, color="#FFFFFF"):
        self.id = id
        self.name = name
        self.parent_group_id = parent_group_id
        self.color = color




class EventManager:

    str_index = {"title", "memo", "metadata"}
    int_index = {"is_all_day", "status", "type", "repeat_group_id"}

    def make_event(event):
        con = get_connection()
        cursor = con.cursor()

        if event.metadata is None:
            event.metadata = {}


        cursor.execute("""
        INSERT INTO event (
            title,
            start_time,
            end_time,
            group_id,
            is_all_day,
            status,
            type,
            repeat_group_id,
            color,
            memo,
            metadata
        )
        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
        """,
        (
            event.title,
            event.start_time,
            event.end_time,
            event.group_id,
            int(event.is_all_day),
            event.status,
            event.type,
            event.repeat_group_id,
            event.color,
            event.memo,
            json.dumps(event.metadata)
        ))

        con.commit()
        con.close()

    def edit_event(event_id, event):
        con = get_connection()
        cursor = con.cursor()

        if event == None:
            cursor.execute("""
            DELETE FROM event
            WHERE id = ?
            """,
            (event_id,))
        else:
            if event.metadata is None:
                event.metadata = {}

            cursor.execute("""
            UPDATE event
            SET title = ?, start_time = ?, end_time = ?, group_id = ?, is_all_day = ?, status = ?, type = ?, repeat_group_id = ?, color = ?, memo = ?, metadata = ?
            WHERE id = ?
            """,
            (
                event.title,
                event.start_time,
                event.end_time,
                event.group_id,
                int(event.is_all_day),
                event.status,
                event.type,
                event.repeat_group_id,
                event.color,
                event.memo,
                json.dumps(event.metadata),
                event_id
            ))

        con.commit()
        con.close()

    def get_event(event_id):
        con = get_connection()
        cursor = con.cursor()

        cursor.execute("""
        SELECT * FROM event
        WHERE id = ?
        """,
        (event_id,))

        row = cursor.fetchone()
        if row is None:
            return None

        event = Event.make(row)

        con.close()
        return event

    def find_event(event, 특징, 형질):

        con = get_connection()
        cursor = con.cursor()

        if(특징 in EventManager.str_index):
            cursor.execute(f"""
                SELECT *
                FROM event
                WHERE {특징} LIKE ?
                """, (f"%{형질}%",))
        elif(특징 in EventManager.int_index):
            cursor.execute(f"""
                SELECT *
                FROM event
                WHERE {특징} = ?
                """, (형질))
        else:
            con.close()
            raise ValueError("Invalid 특징")
            return

        events = [Event.make(row) for row in cursor.fetchall()]

        con.close()
        return events

    def get_events_by_time(start_time, end_time):
        con = get_connection()
        cursor = con.cursor()

        cursor.execute("""
        SELECT *
        FROM event
        WHERE start_time < ? AND end_time > ?
        """,
        (end_time, start_time))

        events = [Event.make(row) for row in cursor.fetchall()]

        con.close()
        return events
    
    def get_events_by_group(group_id):

        group_ids = GroupManager.get_child_groups(group_id)

        if not group_ids:
            return []

        placeholders = ",".join("?" for _ in group_ids)

        con = get_connection()
        cursor = con.cursor()

        cursor.execute(f"""
        SELECT *
        FROM event
        WHERE group_id IN ({placeholders})
        ORDER BY start_time
        """, group_ids)

        events = [Event.make(row) for row in cursor.fetchall()]

        con.close()
        return events

class GroupManager:
    @staticmethod
    def make_group(group):
        con = get_connection()
        cursor = con.cursor()

        if group.metadata is None:
            group.metadata = {}


        cursor.execute("""
        INSERT INTO event_group (
            name,
            parent_group_id,
            color
        )
        VALUES (?, ?, ?)
        """,
        (
            group.name,
            group.parent_group_id,
            group.color
        ))

        con.commit()
        con.close()

    def edit_group(group_id, group):
        con = get_connection()
        cursor = con.cursor()

        if group == None:
            cursor.execute("""
            DELETE FROM event_group
            WHERE id = ?
            """,
            (group_id,))
        else:
            if group.metadata is None:
                group.metadata = {}

            cursor.execute("""
            UPDATE event_group
            SET name = ?, parent_group_id = ?, color = ?
            WHERE id = ?
            """,
            (
                group.name,
                group.parent_group_id,
                group.color,
                group_id
            ))

        con.commit()
        con.close()
    
    def get_group(group_id):
        con = get_connection()
        cursor = con.cursor()

        cursor.execute("""
        SELECT * FROM event_group
        WHERE id = ?
        """,
        (group_id,))

        row = cursor.fetchone()
        if row is None:
            return None

        group = Group(
            id=row["id"],
            name=row["name"],
            parent_group_id=row["parent_group_id"],
            color=row["color"]
        )

        con.close()
        return group
    
    def get_child_groups(group_id):
        con = get_connection()
        cursor = con.cursor()

        cursor.execute(f"""
                SELECT id
                FROM event_group
                WHERE parent_group_id = ?
                """, (group_id,))
    
        child_ids = [row[0] for row in cursor.fetchall()] + [group_id]
        con.close()
        return child_ids
```

```python
PRAGMA foreign_keys = ON;

CREATE TABLE groups (
    id INTEGER PRIMARY KEY,

    name TEXT NOT NULL,

    parent_group_id INTEGER,

    FOREIGN KEY (parent_group_id)
        REFERENCES groups(id)
);

CREATE TABLE event (
    id INTEGER PRIMARY KEY,

    title TEXT NOT NULL,

    description TEXT,

    start_time TEXT NOT NULL,
    end_time TEXT,

    is_all_day INTEGER NOT NULL
        DEFAULT 0
        CHECK (is_all_day IN (0, 1)),

    group_id INTEGER,

    status TEXT NOT NULL
        DEFAULT 'passive'
        CHECK (
            status IN (
                'passive',
                'todo',
                'done'
            )
        ),

    type TEXT NOT NULL
        DEFAULT 'default',

    repeat_group_id INTEGER,

    metadata TEXT
        DEFAULT '{}',

    created_at TEXT NOT NULL
        DEFAULT CURRENT_TIMESTAMP,

    FOREIGN KEY (group_id)
        REFERENCES groups(id),

    CHECK (
        end_time IS NULL
        OR end_time >= start_time
    )
);

CREATE INDEX idx_event_start_time
ON event(start_time);

CREATE INDEX idx_event_group
ON event(group_id);

CREATE INDEX idx_event_group_time
ON event(group_id, start_time);

CREATE INDEX idx_groups_parent
ON groups(parent_group_id);
```
