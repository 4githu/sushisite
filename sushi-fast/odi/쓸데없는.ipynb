````python
from google import genai
from google.genai import types

from pydantic import BaseModel, Field
from typing import List

from dotenv import load_dotenv
import os

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)


class AudienceState(BaseModel):
    E: float = Field(description="-1~1, Engagement")
    V: float = Field(description="-1~1, Evaluative Valence")
    C: float = Field(description="-1~1, Cognitive Clarity")


class AudienceOutput(BaseModel):
    id: int
    state: AudienceState
    currentThought: str
    overallImpression: str


class PresentationOutput(BaseModel):
    presentationSummary: str
    audiences: List[AudienceOutput]


SYSTEM_PROMPT = """
당신은 발표를 듣는 AI 청중 시뮬레이터이다.

입력으로 다음 정보가 제공된다.
- presentationSummary: 지금까지 발표 내용을 누적 요약한 것
- latestSpeech: 발표자가 방금 말한 내용
- audiences: 각 청중의 성격(persona), 현재 상태(state), 현재 생각(currentThought), 발표자에 대한 전체 인상(overallImpression)

당신의 역할은 latestSpeech를 반영하여 각 청중의 내부 상태를 갱신하는 것이다.

평가 기준

E (Engagement)
-1 : 발표에 거의 관심이 없고 집중하지 않음
 0 : 보통 수준
+1 : 매우 집중하고 있음

V (Evaluative Valence)
-1 : 발표를 매우 부정적으로 평가
 0 : 중립
+1 : 매우 긍정적으로 평가

C (Cognitive Clarity)
-1 : 발표를 거의 이해하지 못함
 0 : 부분적으로 이해
+1 : 매우 명확하게 이해

갱신 규칙

- persona를 반영한다.
- 기존 state를 기반으로 자연스럽게 변화시킨다.
- latestSpeech만큼만 변화시킨다.
- 발표가 명확하면 C를 증가시킨다.
- 발표가 흥미롭거나 자신의 관심사이면 E를 증가시킨다.
- 발표 품질이 좋거나 설득력이 높으면 V를 증가시킨다.
- currentThought는 현재 머릿속에서 드는 생각을 한 문장으로 작성한다.
- overallImpression은 발표가 끝났을 때 남는 누적 인상이므로 급격하게 바뀌지 않는다.
- presentationSummary는 latestSpeech를 반영하여 기존 summary를 자연스럽게 갱신한다.
- state의 모든 값은 반드시 [-1, 1] 범위 안에 있어야 한다.

출력은 반드시 JSON만 반환한다.
"""


input_json = """
{
    "presentationSummary": "...",
    "latestSpeech": "...",
    "audiences": [
        {
            "id": 1,
            "persona": {
                "interest":"AI",
                "patience":"high"
            },
            "state":{
                "E":0.3,
                "V":0.2,
                "C":0.1
            },
            "currentThought":"아직은 평범하다.",
            "overallImpression":"무난한 발표."
        }
    ]
}
"""

input_json = """
{
    "presentationSummary": "발표자는 XR 환경에서 발표 연습을 할 때 실제 청중처럼 반응하는 AI 청중 시스템의 필요성을 설명했다. 지금까지 시스템의 목표와 기본 구조를 소개했고, 발표 음성을 분석해 청중 상태를 갱신한다는 흐름을 설명하는 중이다.",
    "latestSpeech": "이 시스템에서는 발표자의 말을 5초 단위로 받아서 텍스트로 변환합니다. 그리고 그 텍스트를 기반으로 각 청중이 얼마나 집중하고 있는지, 발표를 긍정적으로 보는지, 내용을 잘 이해하고 있는지를 계속 업데이트합니다. 예를 들어 설명이 명확하면 이해도가 올라가고, 같은 말이 반복되거나 근거가 부족하면 집중도나 평가가 조금씩 떨어질 수 있습니다.",
    "audiences": [
        {
            "id": 1,
            "persona": {
                "interest": "AI",
                "priorKnowledge": "high",
                "patience": "high",
                "criticalBias": 0.35
            },
            "state": {
                "E": 0.42,
                "V": 0.18,
                "C": 0.51
            },
            "currentThought": "AI 청중이라는 아이디어는 흥미롭고 구조도 어느 정도 이해된다.",
            "overallImpression": "기술적 방향성이 분명하고 꽤 설득력 있는 발표다."
        },
        {
            "id": 2,
            "persona": {
                "interest": "presentation_training",
                "priorKnowledge": "medium",
                "patience": "medium",
                "criticalBias": 0.55
            },
            "state": {
                "E": 0.31,
                "V": 0.05,
                "C": 0.22
            },
            "currentThought": "발표 훈련에 도움이 될 것 같지만 실제로 어떻게 구현되는지는 아직 조금 궁금하다.",
            "overallImpression": "아이디어는 괜찮지만 구체적인 검증 방법이 더 필요해 보인다."
        },
        {
            "id": 3,
            "persona": {
                "interest": "design",
                "priorKnowledge": "low",
                "patience": "medium",
                "criticalBias": 0.25
            },
            "state": {
                "E": 0.12,
                "V": 0.02,
                "C": -0.18
            },
            "currentThought": "무슨 시스템인지는 알겠지만 기술 설명이 조금 어렵게 느껴진다.",
            "overallImpression": "흥미로운 발표지만 아직은 조금 추상적으로 들린다."
        }
    ]
}
"""


response = client.models.generate_content(
    model="gemini-2.5-flash-lite",
    contents=input_json,
    config=types.GenerateContentConfig(
        system_instruction=SYSTEM_PROMPT,
        response_mime_type="application/json",
        response_schema=PresentationOutput,
        temperature=0.3,
        max_output_tokens=2048,
    ),
)

print(response.text)

result = PresentationOutput.model_validate_json(response.text)

print(result)
````

이건 거의 모든 옵션이 적혀 있어서 처음 보면 부담스러워. **네 프로젝트에서 중요한 것**에는 ⭐를 붙여둘게.

---

### 함수

`transcribe_file(...)`

Deepgram의 STT REST API를 이용해 **오디오 또는 비디오 파일을 텍스트로 변환**한다.

---

### 매개변수(Parameters)

⭐ **request**

* 전사할 오디오 데이터.
* `bytes`(파일 바이트) 또는 바이트 스트림을 전달한다.

⭐ **callback**

* 전사가 끝난 후 결과를 받을 URL.
* 긴 작업을 비동기로 처리할 때 사용.

**callback_method**

* callback 요청에 사용할 HTTP 메서드(GET, POST 등).

**extra**

* 응답에 함께 넣어둘 임의의 데이터(Key-Value).

**sentiment**

* 텍스트의 감정(긍정/부정 등)을 분석.

**summarize**

* 전사 내용을 요약.

**tag**

* 요청을 구분하기 위한 태그.

**topics**

* 대화의 주제를 자동으로 추출.

**custom_topic**

* 사용자가 지정한 주제를 우선적으로 탐지.

**custom_topic_mode**

* 사용자 지정 주제만 탐지할지(`strict`), 일반 주제도 함께 탐지할지(`extended`) 설정.

**intents**

* 화자의 의도(Intent)를 분석.

**custom_intent**

* 사용자가 지정한 의도를 우선 탐지.

**custom_intent_mode**

* 의도 탐지 방식을 설정.

**detect_entities**

* 사람 이름, 회사명, 장소 등 개체(Entity)를 추출.

**detect_language**

* 음성의 언어를 자동으로 감지.

**diarize** *(구버전)*

* 화자 분리(누가 말했는지 구분).

**diarize_model**

* 사용할 화자 분리 모델 선택.

**dictation**

* 받아쓰기 모드(구두점 등을 음성으로 말하는 경우에 적합).

**encoding**

* 입력 오디오의 인코딩 형식 지정(PCM, FLAC 등).

⭐ **filler_words**

* `"어"`, `"음"`, `"uh"`, `"um"` 같은 **필러(filler word)를 그대로 전사**할지 여부.
* 네 프로젝트에서는 매우 중요한 옵션.

**keyterm**

* Nova-3 전용.
* 특정 전문 용어나 브랜드명을 더 잘 인식하도록 힌트를 제공.

**keywords**

* 특정 단어나 용어의 인식률을 높이거나 낮춤.

⭐ **language**

* 음성 언어(BCP-47 형식).
* 예: `ko-KR`, `en-US`.

**measurements**

* "3미터" 같은 측정 단위를 약어로 변환.

⭐ **model**

* 사용할 STT 모델.
* 예: `nova-3`.

**multichannel**

* 여러 오디오 채널(좌/우, 마이크 여러 개)을 각각 독립적으로 전사.

**numerals**

* `"삼백"` → `"300"`처럼 숫자를 숫자 형태로 변환.

**paragraphs**

* 결과를 문단 단위로 나눔.

**profanity_filter**

* 욕설을 순화하거나 제거.

⭐ **punctuate**

* 마침표, 쉼표, 대문자 등을 자동으로 추가.

⭐ **redact**

* 개인정보(전화번호, 주민번호 등)를 가림 처리.

**replace**

* 특정 단어나 문장을 다른 표현으로 치환.

**search**

* 특정 단어나 문장을 검색.

⭐ **smart_format**

* 결과를 사람이 읽기 좋게 자동 포맷팅.
* 날짜, 숫자, 전화번호 등을 보기 좋게 변환.

⭐ **utterances**

* 긴 음성을 의미 단위(발화 단위)로 분리.

⭐ **utt_split**

* 몇 초 이상 침묵하면 새로운 발화로 나눌지 설정.

**version**

* 사용할 모델 버전.

**mip_opt_out**

* Deepgram의 모델 개선 프로그램에 데이터를 사용하지 않도록 설정.

**request_options**

* 요청 자체에 대한 추가 설정(타임아웃 등).

---

### 반환값(Returns)

`MediaTranscribeResponse`

* 일반적으로는 **전사 결과(JSON)** 를 반환한다.
* callback 방식을 사용하는 경우에는 **request_id**를 반환한다.

## 네 프로젝트에서 실제로 신경 쓸 옵션

사실 너는 이 정도만 보면 된다.

```python
client.listen.v1.media.transcribe_file(
    request=audio_bytes,
    model="nova-3",
    language="ko-KR",
    filler_words=True,
    punctuate=False,
    smart_format=False,
    utterances=True
)
```

특히 내가 가장 관심 있게 볼 옵션은

* ⭐ `filler_words`
* ⭐ `punctuate`
* ⭐ `smart_format`
* ⭐ `utterances`
* ⭐ `utt_split`

이 다섯 개야.

왜냐하면 발표 평가에서는

* "어...", "음..."을 살려야 하고
* 문장을 AI가 예쁘게 고치면 안 되고
* 어디서 말을 끊었는지도 중요하기 때문이야.

**TL;DR:** 이 함수에는 40개가 넘는 옵션이 있지만, 네 프로젝트에서는 `request`, `model`, `language`, `filler_words`, `punctuate`, `smart_format`, `utterances`, `utt_split` 정도만 알면 충분하다. 나머지는 감정 분석, 요약, 주제 추출, 개인정보 제거 같은 부가 기능이다.
